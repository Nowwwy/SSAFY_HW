```python
import requests

black_list = [
    'Hoeger LLC',
    'Keebler LLC',
    'Yost and Sons',
    'Johns Group',
    'Romaguera-Crona',
]


def censorship(name, company):
    if company in black_list:
        print(f'{company} 소속의 {name} 은/는 등록할 수 없습니다.')
        return False
    else:
        print('이상 없습니다.')
        return True


def create_user(user_list):
    censored_user_list = {}
    for user in user_list:
        name = user['name']
        company = user['company']
        if censorship(name, company):
            censored_user_list.setdefault(company, []).append(name)
    return censored_user_list


user_list = []

for i in range(10):
    API_URL = f'https://jsonplaceholder.typicode.com/users/{i + 1}'
    response = requests.get(API_URL)
    parsed_data = response.json()

    name = parsed_data['name']
    company = parsed_data['company']['name']

    user_list.append({'name': name, 'company': company})

censored_user_list = create_user(user_list)
print(censored_user_list)
```
